```python
from langchain_classic.agents import Tool, initialize_agent, AgentType
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_classic.utilities import DuckDuckGoSearchAPIWrapper
```

```python
llm = ChatGoogleGenerativeAI(model = "gemini-3.1-flash-lite", google_api_key = "YOUR_GOOGLE_API_KEY")
```

```python
search = DuckDuckGoSearchAPIWrapper()
search_tool = Tool(name = "DuckDuckGo Search", func = search.run, description = "For searching via duckduckgo")
```

```python
def simple_calculator(query):
    try:
        return(str(eval(query)))
    except exception as e:
        print(f"Error : {e}")

math_tool = Tool(name = "Math Calculator", func = simple_calculator, description = "Performing math calculations")
```

```python
tools = [search_tool, math_tool]
```

```python
agent = initialize_agent(
    llm = llm,
    tools = tools,
    agent = AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose = True
)
```

```python
query = ""

while query != "bye":
    query = input("Ask your question, If you want to exit type \"bye\":")
    if query == "bye":
        break
    agent.run(query)
```
